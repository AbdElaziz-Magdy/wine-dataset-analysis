# Wine Dataset Analysis

## Project Overview

This project explores a wine dataset using Python and Pandas.

The analysis focuses on data cleaning, missing values, wine varieties, countries, wineries, prices, and wine ratings.

#1.Data Loading


```python
import pandas as pd
import numpy as np
df = pd.read_csv("Wine_data.csv")
df.head()
```

#2.Data Overview

```python
df.shape
```

```python
df.info()
```

#3.Data Cleaning

```python
df.isnull().sum()
```

```python
df.isnull().mean() * 100
```

```python
df = df.drop("Unnamed: 0", axis=1)
df.head()
```

```python
df.duplicated().sum()
```

#4.Statistical Overview

```python
df.describe()
```

#5.Categorical Analysis

```python
df["region_2"].value_counts().head(10)
```

```python
df["country"].value_counts().head(10)
```

```python
df.groupby("country")["points"].mean().sort_values(ascending=False).head(10)
```

```python
df["variety"].value_counts().head(10)
```

#6.Price & Points Analysis

```python
df[["price", "points"]].corr()
```

```python
df.groupby("points")["price"].mean().sort_values(ascending=False).head(10)
```

#7.Varitey Analysis

```python
df.groupby("variety")["points"].mean().sort_values(ascending=False).head(10)
```

```python
df.groupby("variety")["points"].agg(["mean", "count"]).sort_values("mean", ascending=False).head(10)
```

```python
df.groupby("variety")["price"].mean().sort_values(ascending=False).head(10)
```

```python
df.groupby("variety")["price"].agg(["mean", "count"]).sort_values("mean", ascending=False).head(10)
```

#8.Winery Analysis

```python
df["winery"].value_counts().head(10)
```

```python
df.groupby("winery")["price"].mean().sort_values(ascending=False).head(10)
```
